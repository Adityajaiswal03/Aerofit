## 1. **Income**

```python
Income_data=df['Income']
max_income=Income_data.max()
min_income=Income_data.min()
mean_income=Income_data.mean()
print("Max Income= ",max_income,  "\nMin Income= ",min_income, "\nMean Income= ",mean_income)
```

```python
plt.figure(figsize=(8, 6))
plt.boxplot(Income_data)
plt.title('Box plot of Income Data')
plt.ylabel('Income')
plt.show()
```

finding the outliers


```python
Q1_Income = df['Income'].quantile(0.25)
Q3_Income = df['Income'].quantile(0.75)
IQR_Income = Q3_Income - Q1_Income

lower_bound_Income = Q1_Income - 1.5 * IQR_Income
upper_bound_Income = Q3_Income + 1.5 * IQR_Income

outliers_Income = df[(df['Income'] < lower_bound_Income) | (df['Income'] > upper_bound_Income)]

print("Lower Bound =", lower_bound_Income)
print("Upper Bound =", upper_bound_Income)

print("Outliers based on Income:")
outliers_Income
```

```python
sns.histplot(data = df, x = 'Income', bins = 10)
plt.show()
```

```python
sns.kdeplot(data = df, x = 'Income')
plt.show()
```

```python
Income_data_category=pd.cut(df['Income'], bins = [29500 , 50000, 70000, 105000 ],labels=['Low', 'Medium', 'High'])
df['Income_Category'] = Income_data_category
```

```python
sns.countplot(data=df, x='Income_Category')
plt.xlabel('Income_Category')
plt.ylabel('Number of users')
```

```python
sns.countplot(data=df, x='Product', hue='Income_Category')
plt.xlabel('Product')
plt.ylabel('Count')
plt.title('Income Distribution Among Products')
plt.show()
```

### Insight
- Maximum Income: The highest recorded income among individuals is $104,581.

- Minimum Income: The lowest recorded income among individuals is $29,562.

- Mean Income: The average income among individuals is approximately $53,719.58.

- Income Range: The income range, calculated as the difference between the maximum and minimum incomes, is $75,019.

- Lower Bound: The lower bound for identifying potential outliers in the income data, calculated using the interquartile range (IQR) method, is approximately $22,144.88.

- Upper Bound: The upper bound for identifying potential outliers in the income data, calculated using the interquartile range (IQR) method, is approximately $80,581.88.

### Conclusion
It seems that the number of buyers is highest within the income range of 30,000 to 68,000. However, as the income level gradually decreases beyond this range, there is a noticeable decline in the number of buyers.
